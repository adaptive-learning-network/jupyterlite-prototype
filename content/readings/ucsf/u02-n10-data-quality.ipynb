#### What causes poor quality data?

1. Data entry

- Missing data
    - delayed data entry 
    - skipped fields
- Incorrect data 
    - illegibility
    - typos 
    - transcription errors 
    - transposition errors

2. HIS system software 

- Delayed records
- Missing data values and types 
- Duplicate records

#### Informatics Solutions Targeting Data Quality

- Healthcare workers across the country consistently use the same HIS software to collect and report surveillance data
- Data dictionaries standardize data types and formats
- Software applications include 
    - required fields and response options based on data standards
    - data quality checks or validation rules

### Scenario: Martha’s Experience Today

Since the MOH began implementing their 2022 HIS Strategy, Martha received training on and started using a new, standard Community Health Care app. This app replaced several of the other software applications she used to use.

The standard community health information application has several features that support data quality:

- Requires completion of a core set of data fields, such as diagnosis, onset date, residence, and others.

### What is data quality?

Definition
- The level of accuracy and completeness of data in a dataset

### Dimensions of Data Quality

- **Completeness:** are expected records and required fields present?
- **Timeliness:** do reports arrive in time for action?
- **Accuracy:** do values match the event or source record?
