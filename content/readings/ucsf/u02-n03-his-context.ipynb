### Understanding the Country & Health Information System Context

Understand the Health Information System Context

Understanding the HIS context can determine if and how to engage with the software applications and innovations used within a country’s health information system.

The health information system context is:
- fluid, multi-faceted, and ever-changing;
- defined by the political and economic environment, technology infrastructure, and other factors;

### Understanding the HIS Context

- What is happening in the political and economic environment that influences the HIS software applications used at all levels of the system?
- What software applications are currently being used at each level of the system? 
- Which HIS software applications manage the data needed for surveillance and outbreak investigations?
- Who uses these HIS software applications?
